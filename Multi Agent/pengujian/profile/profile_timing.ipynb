```python
# ============================================================
# LANGKAH 1: Import Library & Setup
# ============================================================
import sys
import os
import time
import json
import pandas as pd
from datetime import datetime

# Tambahkan parent folder ke path agar bisa import module Multi Agent
sys.path.insert(0, os.path.abspath('..'))

from agents.profile_agent import run as profile_run
from state import GameState

print("Semua library berhasil diimport!")
print(f"Timestamp: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
# ============================================================
# LANGKAH 2: Definisikan Variasi Usia & Gender
# ============================================================
# 3 kategori usia × 2 gender = 6 variasi profil NPC

USIA_LIST = ["remaja", "dewasa", "orang tua"]
GENDER_LIST = ["pria", "wanita"]
JUMLAH_ITERASI = 5  # Berapa kali setiap variasi dipanggil

variasi_list = []
for usia in USIA_LIST:
    for gender in GENDER_LIST:
        variasi_list.append((usia, gender))

print(f"Total variasi: {len(variasi_list)} (3 usia × 2 gender)")
print(f"Iterasi per variasi: {JUMLAH_ITERASI}")
print(f"Total panggilan profile agent: {len(variasi_list) * JUMLAH_ITERASI}")
print(f"\nDaftar variasi:")
for i, (u, g) in enumerate(variasi_list, 1):
    print(f"  {i}. usia={u}, gender={g}")
```

```python
# ============================================================
# LANGKAH 3: Jalankan Profile Agent & Ukur Waktu
# ============================================================
# Setiap panggilan profile_agent.run() akan diukur waktunya
# menggunakan time.perf_counter() untuk presisi tinggi.

hasil_semua = []
total_panggilan = len(variasi_list) * JUMLAH_ITERASI
panggilan_ke = 0

for usia, gender in variasi_list:
    for iterasi in range(1, JUMLAH_ITERASI + 1):
        panggilan_ke += 1
        
        # Buat state input untuk profile agent
        state: GameState = {
            "usia": usia,
            "gender": gender,
        }
        
        # Catat waktu mulai
        t_start = time.perf_counter()
        
        # Panggil profile agent
        result = profile_run(state)
        
        # Catat waktu selesai
        t_end = time.perf_counter()
        waktu_detik = round(t_end - t_start, 3)
        
        # Simpan hasil
        hasil_semua.append({
            "usia": usia,
            "gender": gender,
            "iterasi": iterasi,
            "nama": result.get("nama", ""),
            "background": result.get("background", ""),
            "masalah_hari_ini": result.get("masalah_hari_ini", ""),
            "ocean_openness": result["ocean"]["openness"],
            "ocean_conscientiousness": result["ocean"]["conscientiousness"],
            "ocean_extraversion": result["ocean"]["extraversion"],
            "ocean_agreeableness": result["ocean"]["agreeableness"],
            "ocean_neuroticism": result["ocean"]["neuroticism"],
            "max_fails": result.get("max_fails", 0),
            "estimasi_waktu_detik": waktu_detik,
        })
        
        print(f"[{panggilan_ke}/{total_panggilan}] usia={usia}, gender={gender}, "
              f"iterasi={iterasi} → {result.get('nama', '?')} "
              f"({waktu_detik}s)")

print(f"\n✅ Selesai! Total {len(hasil_semua)} profile berhasil digenerate.")
```

```python
# ============================================================
# LANGKAH 4: Buat DataFrame & Tampilkan Data
# ============================================================
df = pd.DataFrame(hasil_semua)

# Urutkan kolom agar rapi
kolom_urutan = [
    "usia", "gender", "iterasi", "nama",
    "ocean_openness", "ocean_conscientiousness", "ocean_extraversion",
    "ocean_agreeableness", "ocean_neuroticism",
    "max_fails", "estimasi_waktu_detik",
    "background", "masalah_hari_ini"
]
df = df[kolom_urutan]

print(f"Total data: {len(df)} baris")
print(f"Kolom: {list(df.columns)}")
print(f"\nTipe data:")
print(df.dtypes)
print(f"\n10 data pertama:")
df.head(10)
```

```python
# ============================================================
# LANGKAH 5: Statistik Estimasi Waktu per Kategori
# ============================================================
print("=" * 70)
print("STATISTIK ESTIMASI WAKTU PER KATEGORI (USIA + GENDER)")
print("=" * 70)

summary_kategori = df.groupby(["usia", "gender"]).agg(
    jumlah=("estimasi_waktu_detik", "count"),
    rata_rata_detik=("estimasi_waktu_detik", "mean"),
    std_detik=("estimasi_waktu_detik", "std"),
    min_detik=("estimasi_waktu_detik", "min"),
    max_detik=("estimasi_waktu_detik", "max"),
    median_detik=("estimasi_waktu_detik", "median"),
).round(3)

print(summary_kategori)
print()

# Rata-rata per kategori dalam bentuk tabel pivot
print("=" * 70)
print("RATA-RATA WAKTU (detik) — PIVOT TABLE")
print("=" * 70)
pivot_mean = df.pivot_table(
    values="estimasi_waktu_detik",
    index="usia",
    columns="gender",
    aggfunc="mean"
).round(3)
print(pivot_mean)
```

```python
# ============================================================
# LANGKAH 6: Statistik Estimasi Waktu per Usia & per Gender
# ============================================================

print("=" * 70)
print("RATA-RATA WAKTU PER USIA")
print("=" * 70)
summary_usia = df.groupby("usia").agg(
    jumlah=("estimasi_waktu_detik", "count"),
    rata_rata_detik=("estimasi_waktu_detik", "mean"),
    std_detik=("estimasi_waktu_detik", "std"),
    min_detik=("estimasi_waktu_detik", "min"),
    max_detik=("estimasi_waktu_detik", "max"),
).round(3)
print(summary_usia)

print("\n" + "=" * 70)
print("RATA-RATA WAKTU PER GENDER")
print("=" * 70)
summary_gender = df.groupby("gender").agg(
    jumlah=("estimasi_waktu_detik", "count"),
    rata_rata_detik=("estimasi_waktu_detik", "mean"),
    std_detik=("estimasi_waktu_detik", "std"),
    min_detik=("estimasi_waktu_detik", "min"),
    max_detik=("estimasi_waktu_detik", "max"),
).round(3)
print(summary_gender)
```

```python
# ============================================================
# LANGKAH 7: Tabel Lengkap (Semua Hasil)
# ============================================================
# Tampilkan semua data tanpa kolom background & masalah (terlalu panjang)
kolom_tampil = [c for c in df.columns if c not in ["background", "masalah_hari_ini"]]
print(f"Menampilkan {len(df)} baris data:")
print("-" * 100)
df[kolom_tampil]
```

```python
# ============================================================
# LANGKAH 8: Simpan Hasil ke CSV
# ============================================================
output_path = "hasil_profile_timing.csv"
df.to_csv(output_path, index=False, encoding="utf-8")
print(f"✅ Hasil disimpan ke: {output_path}")
print(f"   Jumlah baris: {len(df)}")
print(f"   Jumlah kolom: {len(df.columns)}")
print(f"\nStatistik global estimasi waktu:")
print(f"   Rata-rata : {df['estimasi_waktu_detik'].mean():.3f} detik")
print(f"   Std dev   : {df['estimasi_waktu_detik'].std():.3f} detik")
print(f"   Min       : {df['estimasi_waktu_detik'].min():.3f} detik")
print(f"   Max       : {df['estimasi_waktu_detik'].max():.3f} detik")
print(f"   Median    : {df['estimasi_waktu_detik'].median():.3f} detik")
```
